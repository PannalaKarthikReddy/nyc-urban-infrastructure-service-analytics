## Executive Findings

### 1. Overall Service Demand

The analysis covers **3,655,040 NYC 311 service requests** recorded during 2025 across **15 agencies**, **874 complaint types**, **59 community boards**, and **5 boroughs**.

The overall average resolution time was **234.61 hours**, while **62.18%** of records with resolution information were resolved within 24 hours.

### 2. Agency Workload

Service demand was concentrated across a small number of agencies.

The **New York City Police Department** recorded **1,717,594 service requests**, followed by the **Department of Housing Preservation and Development** with **774,535** requests.

The Department of Sanitation, Department of Transportation, and Department of Environmental Protection followed with **326,780**, **212,103**, and **197,143** requests respectively.

Workload volume and resolution performance varied substantially across agencies, indicating that service demand and operational resolution patterns are not uniform across municipal functions.

### 3. Complaint-Type Concentration

The most frequent complaint categories were:

* **Illegal Parking:** 564,577 requests
* **Noise - Residential:** 456,599 requests
* **HEAT/HOT WATER:** 315,265 requests
* **Blocked Driveway:** 169,126 requests
* **Noise - Street/Sidewalk:** 168,021 requests

The high volume of parking and noise-related requests contrasts with the substantially longer average resolution time observed for HEAT/HOT WATER requests.

### 4. Borough-Level Variation

Brooklyn recorded the highest number of service requests at **1,003,412**, followed by Queens (**813,632**), the Bronx (**781,563**), and Manhattan (**673,197**).

Average resolution time also varied across boroughs. Manhattan recorded an average of **309.63 hours**, while the Bronx recorded **168.56 hours**.

These figures describe observed differences in the 2025 dataset and should not be interpreted as evidence that geography itself caused differences in resolution performance.

### Analytical Note

The findings above describe observed patterns in the dataset. Differences between agencies, complaint types, or boroughs may reflect differences in service type, workflow, reporting practices, case complexity, or other factors not directly measured in this analysis.

## Resolution Performance Findings

### 1. Agency Resolution Patterns

Average resolution time varied considerably across agencies.

The **Economic Development Corporation** recorded the highest average resolution time at **7,246.39 hours**, followed by the **Department of Education** at **2,857.47 hours** and the **Taxi and Limousine Commission** at **2,320.26 hours**.

The **Department of Housing Preservation and Development**, which handled **774,535 requests**, recorded an average resolution time of **367.99 hours**, with **13.56%** of requests resolved within 24 hours.

These differences should be interpreted in the context of different agency responsibilities, service types, case complexity, and operational workflows.

### 2. Complaint Resolution Patterns

Several complaint categories recorded substantially longer average resolution times than the high-volume categories.

For example, **Noise - Helicopter** recorded an average resolution time of **7,246.39 hours** across **20,049 requests**. **Taxi Complaint** recorded **3,393.02 hours** across **9,380 requests**.

Some of the categories with the highest average resolution times had extremely small volumes, such as **FHV Licensee Complaint** with 3 requests and **DEP Highway Condition** with 4 requests.

Therefore, average resolution time for low-volume complaint categories should be treated as an observation rather than a strong operational conclusion.

### 3. Community Board Variation

Resolution performance also varied across community boards.

Among the boards with the highest average resolution times, **10 QUEENS** recorded **702.15 hours**, followed by **07 MANHATTAN** at **675.63 hours** and **06 MANHATTAN** at **414.46 hours**.

These differences represent observed variation in the 2025 data. Community board-level averages can also be influenced by the mix and complexity of service requests handled within each area.

### Analytical Note

Average resolution time is strongly affected by long-running cases and can therefore be highly skewed. For this reason, the dashboard also reports the percentage of requests resolved within 24 hours rather than relying on average resolution time alone.

## Socioeconomic Findings

### 1. Request Rates and Poverty

The analysis examined **2,205–2,189 eligible census tracts**, depending on the socioeconomic variable.

The Spearman correlation between **service request rate and poverty rate was 0.171**, indicating a weak positive association.

The poverty-quartile analysis shows a similar pattern. Tracts in the lower-poverty quartile had a median request rate of **288.72 requests per 1,000 residents**, compared with **375.17 requests per 1,000 residents** in the higher-poverty quartile.

Median resolution time also increased across the poverty quartiles, from **0.279 days** in the lower-poverty quartile to **0.542 days** in the higher-poverty quartile.

### 2. Request Rates and Household Income

The Spearman correlation between **service request rate and median household income was −0.126**, indicating a weak negative association.

The income-quartile analysis shows that lower-income tracts had a median request rate of **374.94 requests per 1,000 residents**, compared with **296.08 requests per 1,000 residents** in higher-income tracts.

Median resolution time was **0.610 days** in the lower-income quartile and **0.326 days** in the higher-income quartile.

### 3. Resolution Time and Socioeconomic Characteristics

The relationship between socioeconomic characteristics and median resolution time was also weak.

The Spearman correlations were:

| Relationship                            | Spearman ρ |
| --------------------------------------- | ---------: |
| Median resolution vs. household income  |     −0.132 |
| Median resolution vs. poverty rate      |      0.123 |
| Median resolution vs. unemployment rate |      0.093 |

These values indicate relatively weak monotonic associations rather than strong relationships.

### 4. Interpretation

The socioeconomic analysis identifies measurable differences in service request rates and resolution patterns across tract-level socioeconomic groups.

However, these results should **not** be interpreted as evidence that poverty or income causes higher service demand or longer resolution times. The observed relationships may reflect multiple factors, including service composition, population characteristics, reporting behavior, geographic context, agency workflows, and other variables not included in this analysis.

The analysis therefore provides evidence of **association and variation**, not causation.


## Methodology & Limitations

### Data Sources

The project uses 2025 NYC 311 service request data as the primary municipal service dataset.

Additional geographic and socioeconomic analysis incorporates:

* NYC geographic boundary information
* U.S. Census American Community Survey (ACS) 2024 5-year data
* Census TIGER/Line geographic boundaries

### Data Processing

The 2025 NYC 311 dataset was processed through a structured data pipeline covering:

1. Data recovery and monthly consolidation
2. Data profiling
3. Data cleaning and validation
4. Geographic reconciliation
5. Census tract spatial analysis
6. Socioeconomic enrichment
7. SQL-based analytical analysis
8. Power BI dashboard preparation

The final validated dataset contains **3,655,040 service requests**.

### Geographic Analysis

Coordinate-valid service requests were spatially assigned to Census tracts using geographic boundary data.

The socioeconomic analysis included population, median household income, poverty rate, unemployment rate, housing units, service request rates, and resolution metrics.

Tracts with populations below 1,000 were excluded from the sensitivity-filtered socioeconomic analysis to reduce instability from very small population denominators.

### Resolution Metrics

Resolution time was calculated from service request creation and closure timestamps.

Both average resolution time and the percentage of requests resolved within 24 hours were examined because resolution times are strongly right-skewed and can be influenced by long-running cases.

### Statistical Analysis

Pearson and Spearman correlation measures were used to examine relationships between service request rates, resolution time, and socioeconomic characteristics.

Correlation results describe statistical association and **do not establish causation**.

### Limitations

Several limitations should be considered when interpreting the results:

* NYC 311 records represent reported service requests rather than all underlying conditions in the city.
* Resolution time may be influenced by agency workflows, case complexity, administrative processes, and reporting practices.
* Average resolution time can be heavily affected by unusually long-running cases.
* Socioeconomic variables are measured at the Census tract level and should not be interpreted as individual-level characteristics.
* Some socioeconomic variables contain incomplete coverage.
* Extremely small complaint categories can produce unstable average resolution times.
* Geographic analyses depend on the availability and quality of location information.
* Observed correlations should not be interpreted as causal relationships.

### Reproducibility

All major analytical stages were completed using documented notebooks, SQL analysis, validated summary datasets, and a Power BI dashboard.

The project is structured so that the analytical workflow can be reviewed and reproduced from the source data-processing stages through the final dashboard.
