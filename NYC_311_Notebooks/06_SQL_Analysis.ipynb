# NYC 311 2025 — SQL Analysis Preparation

This notebook prepares the validated NYC 311 analytical dataset for MySQL-based analysis.

**Analysis period:** January–December 2025

The analytical SQL queries were executed separately in MySQL using the prepared database table.


## 1. Prepare MySQL-Ready Dataset



```python
from google.colab import drive

drive.mount("/content/drive")
```

```python
# STEP 35 — PREPARE MYSQL-READY DATASET

import pandas as pd
import os

SOURCE_PATH = (
    "/content/drive/MyDrive/"
    "NYC_311_2025_BACKUP/"
    "NYC_311_2025_FINAL.parquet"
)

MYSQL_OUTPUT = (
    "/content/drive/MyDrive/"
    "NYC_311_2025_BACKUP/"
    "NYC_311_2025_MYSQL.csv"
)

df = pd.read_parquet(SOURCE_PATH)

mysql_columns = [
    "service_request_id",
    "created_at",
    "closed_at",
    "resolution_hours",
    "resolution_days",
    "resolution_quality",
    "agency",
    "complaint_type",
    "descriptor",
    "status",
    "resolution_description",
    "incident_zip",
    "borough",
    "community_board",
    "community_board_number",
    "community_board_borough",
    "latitude",
    "longitude",
    "geo_quality_final"
]

mysql_df = df[mysql_columns].copy()

# Convert datetime columns to MySQL-friendly format
mysql_df["created_at"] = pd.to_datetime(
    mysql_df["created_at"],
    errors="coerce"
).dt.strftime("%Y-%m-%d %H:%M:%S")

mysql_df["closed_at"] = pd.to_datetime(
    mysql_df["closed_at"],
    errors="coerce"
).dt.strftime("%Y-%m-%d %H:%M:%S")

# Replace pandas missing values with empty strings
mysql_df = mysql_df.where(
    pd.notna(mysql_df),
    ""
)

mysql_df.to_csv(
    MYSQL_OUTPUT,
    index=False
)

print("=" * 70)
print("MYSQL DATASET PREPARATION")
print("=" * 70)

print("Rows:", len(mysql_df))
print("Columns:", len(mysql_df.columns))
print("Output:", MYSQL_OUTPUT)
print(
    "File size:",
    f"{os.path.getsize(MYSQL_OUTPUT) / (1024**2):.2f} MB"
)

print()
print("Columns:")
for i, col in enumerate(mysql_df.columns, 1):
    print(f"{i:2}. {col}")
```
